```python
def Penultimo(Q:LinkedQueue):
    # Calculamos el índice del elemento que queremos atrapar.
    # Usamos len(Q) - 2 porque queremos el PENÚLTIMO. (len(Q) - 1 sería el último)
    n = len(Q) - 2 
    
    # Hacemos un ciclo que se repita la misma cantidad de veces que el tamaño de la cola.
    # Esto asegura que le daremos una vuelta completa a todos los vagones.
    for i in range(len(Q)):
        # Sacamos el elemento que está al frente de la cola (head)
        x = Q.dequeue()
        
        # Si el contador 'i' llega a la posición del penúltimo elemento...
        if i == n:
            # ... guardamos ese elemento en la variable 'y' para no perderlo
            y = x
            
        # Volvemos a meter el mismo elemento al final de la cola (tail).
        # Al hacer esto len(Q) veces, la cola vuelve a quedar en su orden original.
        Q.enqueue(x)
        
    # Una vez terminada la vuelta completa, devolvemos el valor que atrapamos
    return y
```

```python
def Penultimo1(Q:LinkedQueue):
    # 1. Punto de partida: 
    # 'x' se convierte en nuestro "cursor" o apuntador, empezando en el primer vagón (head).
    x = Q._head
    
    # 2. El ciclo de pasos exactos:
    # len(Q) nos da el total de elementos. 
    # Al restarle 2 (len(Q) - 2), calculamos exactamente cuántos "saltos" 
    # necesitamos dar para caer en el penúltimo vagón.
    for i in range(len(Q)-2):
        # 3. El salto:
        # En cada iteración, 'x' deja de ser el vagón actual y pasa a ser el siguiente (_next).
        x = x._next
        
    # 4. El final del viaje:
    # Cuando el ciclo 'for' termina sus saltos, 'x' quedó estacionado 
    # justo en el penúltimo nodo. Así que extraemos y devolvemos su valor (_element).
    return x._element
```

```python
def concatenar_privado(Q1, Q2):
    # Si la Cola 2 está vacía, no hay nada que hacer
    if Q2.is_empty():
        return
        
    # Si la Cola 1 está vacía, la Cola 1 se convierte en la Cola 2
    if Q1.is_empty():
        Q1._head = Q2._head
        Q1._tail = Q2._tail
    else:
        # Aquí está la magia de los punteros:
        # 1. El 'next' del tail de Q1 apunta al head de Q2
        Q1._tail._next = Q2._head 
        # 2. El nuevo tail de Q1 pasa a ser el tail de Q2
        Q1._tail = Q2._tail       
    
    # Actualizamos el tamaño total (si tu clase lleva un contador '_size')
    Q1._size += Q2._size 
    
    # Opcional pero recomendado: Vaciar Q2 para que no queden referencias "fantasma"
    Q2._head = None
    Q2._tail = None
    Q2._size = 0
```

```python
def concatenar_publico(Q1, Q2):
    # Mientras la Cola 2 tenga elementos
    while not Q2.is_empty():
        # Sacamos el primero de Q2
        elemento = Q2.dequeue()
        # Lo metemos al final de Q1
        Q1.enqueue(elemento)
```

```python
def SobreOtro(Q1, Q2):
    '''Se pone el Q2 debajo del Q1'''
    
    # Usamos un ciclo basado en la cantidad de elementos de Q2.
    # (También podrías usar un "while not Q2.is_empty():" si lo prefieres)
    for i in range(len(Q2)):
        
        # 1. Sacamos al que está al frente de la fila Q2
        x = Q2.dequeue()
        
        # 2. Lo mandamos al final de la fila Q1
        Q1.enqueue(x)
```

```python
def SobreOtro_privado(Q1:LinkedQueue, Q2:LinkedQueue):
    '''Se pone el Q2 debajo del Q1'''
    
    # 1. Conectamos el viejo final de Q1 con el inicio de Q2
    Q1._tail._next = Q2._head
    
    # 2. ¡LA LÍNEA CLAVE! Actualizamos el final de Q1
    Q1._tail = Q2._tail 
    
    # 3. Actualizamos el tamaño total
    Q1._size += len(Q2)
```
