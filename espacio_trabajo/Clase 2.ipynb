```python
class ColaPrioridad:
    def __init__(self):
        self.info = []  # Guardar la información en pares (key, value)
 
    def add(self, key, value):
        '''Adiciona a la cola de prioridad un par (key, value)
        '''
        self.info.append((key, value))
 
    def __len__(self):
        return len(self.info)
   
    def is_empty(self):
        'Retorna True si el PQ es vacio'
        return len(self.info) == 0
 
    def min(self):
        minimo = min([i[0] for i in self.info])
        contador = 0
        for i in self.info:
            if i[0] == minimo:
                return self.info.pop(contador)
            contador += 1
```

```python
import random

# Crear una instancia de ColaPrioridad
cola = ColaPrioridad()

# Agregar 1000 entradas con claves aleatorias
for i in range(1000):
    key = random.randint(1, 10000)  # Claves aleatorias entre 1 y 10000
    value = f"entrada_{i}"  # Valores simples para cada entrada
    cola.add(key, value)

print(f"Cola de prioridad construida con {len(cola)} entradas.")
```
