```python
class DiccionarioSimpleHash():
    def __init__(self, size):
        self._size = size
        self._data = [None] * size
        self._n = 0  # número de elementos

    def _hash(self, key):
        return hash(key) % self._size

    def __getitem__(self, key):
        index = self._hash(key)

        if self._data[index] is None:
            raise KeyError(key)

        k, v = self._data[index]
        return v

    def __setitem__(self, key, value):
        index = self._hash(key)

        if self._data[index] is None:
            self._n += 1

        self._data[index] = (key, value)

    def __delitem__(self, key):
        index = self._hash(key)

        if self._data[index] is None:
            raise KeyError(key)

        self._data[index] = None
        self._n -= 1

    def __len__(self):
        return self._n

    def __iter__(self):
        for item in self._data:
            if item is not None:
                yield item[0]
```

```python
class HeapPriorityQueue1(HeapPriorityQueue):
    def _upheap(self, j, cont = 0):
        parent = self._parent(j)
        if j > 0 and self._data[j] < self._data[parent]:
            cont += 1
            self._swap(j, parent)
            return self._upheap(parent, cont)
        else:
            return cont  
    
    def add(self, key, value):
        """Add a key-value pair to the priority queue."""
        self._data.append(self._Item(key, value))
        return self._upheap(len(self._data) - 1)            # upheap newly added position
```

```python
class ColaPrioridad:
    def __init__(self):
        self.info = []  # Guardar la información en pares (key, value)

    def add(self, key, value):
        '''Adiciona a la cola de prioridad un par (key, value)
        '''
        self.info.append((key, value))

    def __len__(self):
        return len(self.info)
    
    def is_empty(self):
        'Retorna True si el PQ es vacio'
        return len(self.info) == 0

    def remove_min(self):
        minimo = min([i[0] for i in self.info])
        contador = 0
        for i in self.info:
            if i[0] == minimo:
                return self.info.pop(contador)
            contador += 1
```
