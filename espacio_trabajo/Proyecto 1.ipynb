## **Función Principal**

```python
def Notacion_Polaca(exp):
    pila = ArrayStack()

    if isinstance(exp, str):
        caracteres = exp.split()
    elif isinstance(exp, ArrayStack):
        pilatemp = ArrayStack()
        while not exp.is_empty():
            pilatemp.push(exp.pop())
            
        def extraer_de_pila():
            while not pilatemp.is_empty():
                yield pilatemp.pop()
        caracteres = extraer_de_pila()
    else:
        raise TypeError("Tipo de entrada inválido")

    for caracter in caracteres:
        if caracter in "+-*/^":
            if len(pila) < 2:
                print("Error: faltan valores")
                return
            
            num2 = pila.pop()
            num1 = pila.pop()
            
            if caracter == '+':
                rest = num1 + num2
            elif caracter == '-':
                rest = num1 - num2
            elif caracter == '*':
                rest = num1 * num2
            elif caracter == '/':
                rest = num1 / num2
            elif caracter == '^':
                rest = num1 ** num2
                
            pila.push(rest)
        else:
            try:
                numero = float(caracter)
                if numero.is_integer():
                    numero = int(numero)
                pila.push(numero)
            except ValueError:
                print("Error: operador inválido")
                return

    if len(pila) == 0:
        print("Error: faltan valores")
        return
    if len(pila) > 1:
        print("Error: demasiados valores")
        return

    rest_final = pila.pop()
    
    if isinstance(rest_final, float) and rest_final.is_integer():
         rest_final = int(rest_final)
         
    return rest_final



    
```

## **Función Extra**

```python
def infija_a_postfija(exp):
    if not isinstance(exp, str):
        return "Error: La expresión a convertir debe ser una cadena de texto"

    for caracter in '()+-*/^':
        exp = exp.replace(caracter, ' ' + caracter + ' ')

    caracteres = exp.split()
    pila = ArrayStack()

    for caracter in caracteres:
        if caracter == '(':
            pila.push('(')
        elif caracter == ')':
            num2 = pila.pop()
            operador = pila.pop()
            num1 = pila.pop()
            pila.pop()
            
            rest = num1 + ' ' + num2 + ' ' + operador
            pila.push(rest)
        else:
            pila.push(caracter)

    rest_final = pila.pop()
    return rest_final
```

```python
d = ((5+2)*(8-3))
infija_a_postfija(d)
```
