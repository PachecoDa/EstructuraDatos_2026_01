```python
def numero_hojas(BT):
    """Cuenta el número total de hojas en el LinkedBinaryTree."""
    if BT.is_empty():
        return 0
    return _contar_recursivo(BT, BT.root())

def _contar_recursivo(BT, p):
    """Función de apoyo para recorrer el árbol."""
    if BT.is_leaf(p):
        return 1  # Encontramos una hoja
    else:
        # Sumamos las hojas del subárbol izquierdo y del derecho
        total = 0
        if BT.left(p) is not None:
            total += _contar_recursivo(BT, BT.left(p))
        if BT.right(p) is not None:
            total += _contar_recursivo(BT, BT.right(p))
        return total

# Ejemplo de uso con tu árbol BT:
print(f"El número de hojas es: {numero_hojas(BT)}")
```
